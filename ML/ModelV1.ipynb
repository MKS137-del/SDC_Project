```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("Data_Train.csv")
df = data.copy()
```

```python
def fillNa(col):
  df[col] = df[col].fillna(df[col].median)

def checkCounts(col):
  print(df[col].value_counts(dropna = False))

def checkUnique(col):
  print(df[col].unique())

def toNumeric(col):
  df[col] = pd.to_numeric(df[col],errors="coerce")
```

```python
df.columns = df.columns.str.lower()
```
