```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("Data_Train.csv")
df = data.copy()
```

```python
def fillNA(col):
  df[col] = df[col].fillna("Unknown")

def checkCounts(col):
  print(df[col].value_counts(dropna = False))

def checkUnique(col):
  print(df[col].unique())

def toNumeric(col):
  df[col] = pd.to_numeric(df[col],errors="coerce")
```

```python
df.columns = df.columns.str.lower()
```

Handing data and time columns

```python
df["date_of_journey"] = pd.to_datetime(df["date_of_journey"],format='%d/%m/%Y')
```

```python
df["day_of_journey"] = df["date_of_journey"].dt.day
df["month_of_journey"] = df["date_of_journey"].dt.month
```

```python
df[["day_of_journey","month_of_journey"]].head()
```

```python
df = df.drop(columns = "date_of_journey")
```

```python
checkCounts("day_of_journey")
```

```python
# 'date_of_journey', 'dep_time', 'arrival_time', 'duration'
checkCounts("dep_time")
```

```python
df["dep_time"] = pd.to_datetime(df["dep_time"])
```

```python
df["dep_hour"] = df["dep_time"].dt.hour
df["dep_minute"] = df["dep_time"].dt.minute
```

```python
df.drop('dep_time',axis=1,inplace=True)
```

```python
checkCounts("arrival_time")
```

```python
# Check whether arrival is on the next day
df['arrival_next_day'] = df['arrival_time'].str.contains(' ', na=False).astype(int)

# Extract time only
arrival = df['arrival_time'].str.split().str[0]

# Extract hour and minute
df['arrival_hour'] = pd.to_datetime(
    arrival, format='%H:%M'
).dt.hour

df['arrival_minute'] = pd.to_datetime(
    arrival, format='%H:%M'
).dt.minute

# Drop original column
df.drop('arrival_time', axis=1, inplace=True)
```

```python
checkCounts("duration")
```

```python
checkUnique("duration")
```

```python
def convert_duration(duration):
  hours = 0
  minutes = 0

  if 'h' in duration:
    hours = int(duration.split('h')[0])

  if 'm' in duration:
    minutes = int(duration.split('m')[0].split()[-1])

  return hours*60 + minutes


df['duration'] = df['duration'].apply(convert_duration)

df.rename(columns={'duration': 'duration_minutes'}, inplace=True)
```
